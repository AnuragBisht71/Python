```python
## Assignment 8:

total = 0

while True:
    num = float(input("Enter a number:"))
    if num == 0:
        break
    total += num
print(f"The sum of number are {total}")
```

```python
## Assignment 11:

num = int(input("Enter a number:"))

for i in range(1, num+1):
    if i % 2 == 0:
        print(f"{i} is a even number")
```

```python
## Assignment 12:

fact = 1
i = 1
num = int(input("Enter a number:"))

while i <= num:
    fact = fact * i
    i = i +1
print(f"{fact}")
```

```python
## Assignment 13:

sum = 0
num = int(input("Enter a number"))

while num > 0:
    digit = num % 10
    sum += digit
    num = num // 10
    
print(f"The sum of digits is {sum}")
```
