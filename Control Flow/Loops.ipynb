```python
## For Loop

range(5)

for i in range(5):
    print(i)
```

```python
for i in range(1,6):
    print(i)
```

```python
## Range (Start, Stop, Step)

for i in range(1, 10, 2):
    print(i)
```

```python
for i in range(10,1,-1):
    print(i)
```

```python
for i in range(10, 1, -2):
    print(i)
```

```python
## Strings in loop

str="Anurag Bisht"

for i in str:
    print(i)
```

```python
## While Loop

count = 0
while count<5:
    print(count)
    count = count + 1
```

```python
## Loop Control Statement (Break, Continue, Pass)

## Break

for i in range(10):
    if i == 5:
        break
    print(i)
```

```python
## Continue

for i in range(10):
    if i%2 == 0:
        # print(f"{i} is a even number")
        continue
    print(f"{i} is a odd number")
```

```python
## Pass

for i in range(5):
    if i == 3:
        # print("The number is ", i)
        pass
    print(i)
```

```python
## Nested Loops

for i in range(3): ## 0, 1, 2
    for j in range(2): ## 0, 1
        print(f"i:{i} and j:{j}")
```

```python
## Calculating the sum of first 10 natural numbers using for and while loop

n = 10
sum=0
count = 1

while count <= n:
    sum = sum + count
    count = count + 1
    
print(f"The sum of 10 natural numbers is {sum}")
```

```python
sum = 0

for i in range(11):
    sum = sum + i

print(f"The sum of first 10 natural number is {sum}")
```

```python
## Prime numbers between 1 and 100

for i in range(1,101):
    if i > 1:
        for j in range(2, i):
            if i % j == 0:
                break
        else:
            print(f"{i} is a prime number")
```

## Conclusion:

Loops are powerful constructs in python that allows you to execute a block of code multiple times. By understanding and using fo and while loops, along with loops control statements like break, continue, and pass you can handle a wide range of programming tasks efficiently
