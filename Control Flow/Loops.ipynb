```python
## For Loop

range(5)

for i in range(5):
    print(i)
```

```python
for i in range(1,6):
    print(i)
```

```python
## Range (Start, Stop, Step)

for i in range(1, 10, 2):
    print(i)
```

```python
for i in range(10,1,-1):
    print(i)
```

```python
for i in range(10, 1, -2):
    print(i)
```

```python
## Strings in loop

str="Anurag Bisht"

for i in str:
    print(i)
```
