```python
## IF Statement


age=18

if age>=15:
    print("Your are eligible!!")
```

```python
## ELSE Statement


age=20

if age>20:
    print("Your are eligible!!")
else:
    print("Your are not eligible!!")
```

```python
## ELIF Statement

age=21

if age<12:
    print("You are a child")
elif age<18:
    print("You are a teenager")
else:
    print("You are an adult")
```

```python
## Nested Conditional Statement

## Number is even or odd

num=int(input("Write your number"))

if num>0:
    print("The number is positive")
    if num%2==0:
        print("The number is even")
    else:
        print("The number is odd")

else:
    print("The number is zero or negative")
```

```python
## Practical Example

## Leap Year

year=int(input("Write a year!!"))

if year%4==0:
    if year%100==0:
        if year%400==0:
            print(year, "is a Leap Year")
        else:
            print(year, "is not a Leap Year")
    else:
        print(year, "is a Leap Year")
        
else:
    print(year, "is not a Leap Year")
```

```python
## Calculator using If, Else and Else Statement

num1=int(input("First number"))
num2=int(input("Second number"))
operator=input("Write a operator (+, -, *, /):")

if operator == "+" :
    print(num1+num2)
elif operator == "-" :
    print(num1-num2)
elif operator == "*" :
    print(num1*num2)
elif operator == "/" :
    if num2 != 0:
        print(num1/num2)
    else:
        print("Error, Division by Zero")
else:
    print("Invaild operator")
```
