Arithmetic Operations

```python
a=10
b=5

add_result=a+b
sub_result=a-b
mult_result=a*b
div_result=a/b
modulus_result=a%b
floor_div_result=a//b
exponent_result=a**b

print(add_result)
print(sub_result)
print(mult_result)
print(div_result)
print(modulus_result)
print(floor_div_result)
print(exponent_result)
```

Comparison Operations

```python
## Equal To '=='

a=10
b=10

print(a==b)


str1="Anurag"
str2="anurag"

print(str1==str2)
```

```python
## Not Equal To '!='

str1="Anurag"
str2="anurag"

str1!=str2
```

```python
## Greater Than '>' and Less Than '<'

num1=55
num2=45

print(num1>num2)
print(num1<num2)
```

```python
## Greater Than Equal To '>='  and  Less Than Equal To '<='

n1=55
n2=55

print(n1>=n2)
print(n1<=n2)
```

Logical Operations {And, Or, Not}

```python
## 'And' Operation

X=True
Y=False

print(X and Y)

X1=True
Y1=True

print(X1 and Y1)
```

```python
## 'Or' Operation

X=True
Y=True

print(X and Y)

X1=False
Y1=False

print(X1 and Y1)
```

```python
## 'Not' Operation

X=True

print(not X)
```

```python
## Calculator


num1=float(input("What is the first number"))
num2=float(input("What is the second number"))

add=num1+num2
sub=num1-num2
multi=num1*num2
div=num1/num2
floor_div=num1//num2
modulus=num1%num2
exponent=num1**num2


print(add, ": Addition")
print(sub, ": Subtraction")
print(multi, ": Multiplication")
print(div, ": Division")
print(floor_div, ": Floor Division")
print(modulus, ": Modulus")
print(exponent, ": Exponent")
```
