Syntax & Semantics

```python
# Syntax --> Variables

name="Anurag"
Name="Bisht"

print(name)
print(Name)
```

```python
# Syntax --> Indentation

age=32
if age>30:
    print(age)

print(age)
```

```python
# Syntax --> Line Continuation

total=1+2+3+4+5+6+7+\
    4+5+6
    
print(total)
```

```python
# Syntax --> Multiple Statement

x=5;y=10;z=x+y
print(z)
```

```python
# Variable assignment --> Type Inference

age=32
type(age)

variable=10
print(type(variable))

variable=32
print(type(variable))
variable="Anurag"
print(type(variable))
```

```python
# Code Example of Indentation

if True:
    print("Correct indentation")
    if False:
        print("This will not print")
    print("This will print")
print("Outside the if block")
```
